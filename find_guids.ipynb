# Find MemoQ GUIDs

This notebook helps you find the GUIDs for your Translation Memories (TMs) and Term Bases (TBs) in MemoQ.

**You'll need:**
- MemoQ server running and accessible
- Your `.env` file configured with credentials


## 1. Import Required Modules


```python
from memoq_client import MemoQClient, MemoQConfig
from dotenv import load_dotenv

# Load environment variables
load_dotenv()

print("✅ Imports successful")
```

## 2. Connect to MemoQ Server


```python
print("🔍 Connecting to MemoQ server...")
print("=" * 40)

try:
    # Initialize client from .env
    config = MemoQConfig.from_env()
    client = MemoQClient(config)
    
    # Display configuration
    print(f"Server URL: {config.base_url}")
    print(f"Username: {config.username}")
    print(f"Timeout: {config.timeout}s")
    
    # Login
    client.login()
    print("\n✅ Connected to MemoQ server successfully!")
    
except Exception as e:
    print(f"\n❌ Connection failed: {e}")
    print("\nMake sure:")
    print("  - MemoQ server is running")
    print("  - Your .env file has correct credentials")
    print("  - Server is accessible from your network")
```

## 3. Find Your Specific TM and TB GUIDs


```python
print("🎯 Finding your specific resources:")
print("-" * 50)

# Your TM and TB names (update these if needed)
tm_name = "SV_test_TM_Finance"
tb_name = "Finance_TB"

try:
    # Find TM
    tm_guid = client.get_tm_guid_by_name(tm_name, "en", "es")
    
    if tm_guid:
        print(f"✅ Found TM '{tm_name}':")
        print(f"   GUID: {tm_guid}")
    else:
        print(f"❌ TM '{tm_name}' not found")
        print("   Run the cell below to list all available TMs")
    
    # Find TB
    tb_guid = client.get_tb_guid_by_name(tb_name, ["en", "es"])
    
    if tb_guid:
        print(f"\n✅ Found TB '{tb_name}':")
        print(f"   GUID: {tb_guid}")
    else:
        print(f"\n❌ TB '{tb_name}' not found")
        print("   Run the cell below to list all available TBs")
    
    # Generate .env lines
    print("\n📝 Add these to your .env file:")
    print("=" * 50)
    if tm_guid:
        print(f"MEMOQ_TM_GUID={tm_guid}")
    if tb_guid:
        print(f"MEMOQ_TB_GUID={tb_guid}")
    print("=" * 50)
    
except Exception as e:
    print(f"❌ Error: {e}")
```
